## 3. Evaluación del Modelo


```python
from ultralytics import YOLO
import torch

# Cargar modelo entrenado
model = YOLO("runs/detect/futbol_detector_cpu/weights/best.pt")  # Ajusta la ruta

# Evaluar el modelo
metrics = model.val(
    data="M:/GitHub Clase/Tareas-y-Ejercicios-de-Clase/PRIA/YOLO/futbol/data.yaml",
    split='val',
    conf=0.5,
    iou=0.6,
    device='cpu'  # o '0' si usas GPU
)

# Métricas actualizadas (YOLOv8 8.x+)
print("\n📊 Métricas de Evaluación:")
print(f"Precisión (mAP@0.5): {metrics.box.map:.4f}")        # mAP a IoU 0.5
print(f"mAP@0.5-0.95: {metrics.box.map50:.4f}")            # mAP promedio de 0.5 a 0.95 IoU
print(f"Precisión: {metrics.box.mp:.4f}")             # Precisión
print(f"Recuperación: {metrics.box.mr:.4f}")          # Recall
print(f"F1-Score: {2 * (metrics.box.mp * metrics.box.mr) / (metrics.box.mp + metrics.box.mr):.4f}")
```

## 7. Análisis Avanzado de Métricas


```python
from ultralytics.utils.plotting import plot_results
import pandas as pd
import matplotlib.pyplot as plt

# Cargar datos de entrenamiento (ajusta la ruta)
results_path = "runs/detect/futbol_detector_cpu/results.csv"
results = pd.read_csv(results_path)

# Graficar métricas clave
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.plot(results['epoch'], results['metrics/precision(B)'], label='Precisión')
plt.title('Precisión por Época')
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(results['epoch'], results['metrics/recall(B)'], label='Recall', color='orange')
plt.title('Recall por Época')
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(results['epoch'], results['metrics/mAP50(B)'], label='mAP@0.5', color='green')
plt.title('mAP@0.5 por Época')
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(results['epoch'], results['val/box_loss'], label='Pérdida', color='red')
plt.title('Pérdida en Validación')
plt.legend()

plt.tight_layout()
plt.savefig('metricas_entrenamiento.png')  # Guardar gráfico
plt.show()
```

```python
from ultralytics import YOLO
import cv2

# 1. Cargar modelo con ajustes especiales para deportes
model = YOLO("runs/detect/futbol_detector_cpu/weights/best.pt")
model.overrides['conf'] = 0.35  # Reducir umbral para detectar más jugadores
model.overrides['iou'] = 0.45   # Evitar cajas duplicadas

# 2. Procesar video con parámetros optimizados
video_path = "M:/GitHub Clase/Tareas-y-Ejercicios-de-Clase/PRIA/YOLO/futbol"
cap = cv2.VideoCapture(video_path)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    # Ajustes para detectar jugadores pequeños:
    results = model.track(
        frame,
        imgsz=1280,       # Mayor resolución
        persist=True,     # Mantener IDs entre frames
        tracker="bytetrack.yaml",
        augment=True      # Aumentación en tiempo real
    )
    
    # Visualización mejorada
    annotated_frame = results[0].plot(
        line_width=2,     # Grosor de las cajas
        font_size=0.8,    # Tamaño de texto
        labels=True,      # Mostrar etiquetas
        boxes=True        # Mostrar cajas
    )
    
    cv2.imshow('Detección', annotated_frame)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
