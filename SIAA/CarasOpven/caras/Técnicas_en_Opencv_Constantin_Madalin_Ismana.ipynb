<h1>Paso 1: importar el paquete OpenCV<h1>

<p>Ahora, vamos a importar OpenCV e introducir la ruta de la imagen de entrada con las siguientes líneas de código:<p>

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os
```

<p>Ahora vamos a  introducir la ruta de la imagen de entrada con las siguientes líneas de código:<p>

```python
imagePath = 'Made.jpg'
```

<h1>Paso 2: leer la imagen<h1>

<p>A continuación, tenemos que leer la imagen con la función imread() de OpenCV:<p>

```python
img = cv2.imread(imagePath)
```

<p>Vamos a imprimir las dimensiones de esta matriz:<p>

```python
img.shape
```

<h1>Paso 3: convertir la imagen a escala de grises<h1>

<p>Para mejorar la eficiencia computacional, primero tenemos que convertir esta imagen a escala de grises antes de realizar en ella la detección de caras:<p>

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

<p>Examinemos ahora las dimensiones de esta imagen en escala de grises:<p>

```python
gray_image.shape
```

<h1>Paso 4: cargar el clasificador<h1>

<p>Carguemos el clasificador Haar Cascade preentrenado que está incorporado en OpenCV:<p>

```python
face_classifier = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
```

<h1>Paso 5: realizar la detección facial<h1>

<p>Paso 5: realizar la detección facial<p>

```python
face = face_classifier.detectMultiScale(
    gray_image, scaleFactor=1.1, minNeighbors=6, minSize=(40, 40)
)
```

<h1>Paso 6: dibujar un cuadro delimitador<h1>

<p>Ahora que el modelo ha detectado las caras dentro de la imagen, vamos a ejecutar las siguientes líneas de código para crear un cuadro delimitador alrededor de estas caras:<p>

```python
for (x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 4)
```

<h1>Paso 7: visualizar la imagen<h1>

<p>Para visualizar la imagen con las caras detectadas, primero tenemos que convertir la imagen del formato BGR a RGB:<p>

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(20,10))
plt.imshow(img_rgb)
plt.axis('off')
```

<h1>Fotos de Monedas<h1>

```python
# Cargar la imagen
imagePath = "moneda1.jpg"
img = cv2.imread(imagePath)

# Convertir a escala de grises
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Aplicar desenfoque para reducir el ruido
gray_blurred = cv2.medianBlur(gray_image, 5)

# Detectar círculos usando la Transformada de Hough
circles = cv2.HoughCircles(
    gray_blurred, cv2.HOUGH_GRADIENT, dp=1.3, minDist=95,
    param1=50, param2=30, minRadius=10, maxRadius=50
)

# Si se detectan círculos, dibujarlos
if circles is not None:
    circles = np.round(circles[0, :]).astype("int")
    for (x, y, r) in circles:
        cv2.circle(img, (x, y), r, (0, 255, 0), 5)  # Cambiar grosor de línea a 2

# Convertir a RGB para mostrar correctamente con matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Mostrar la imagen
plt.figure(figsize=(10, 8))
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

<h1>Reconocimiento facial<h1>

```python
# Rutas de archivos
image_folder = 'C:/Users/constantin.madalin.i/Desktop/Tareas-y-Ejercicios-de-Clase/SIAA/CarasOpven/caras'  # Carpeta donde están las imágenes
video_path = 'C:/Users/constantin.madalin.i/Desktop/Tareas-y-Ejercicios-de-Clase/SIAA/CarasOpven/caras/niki1.mp4'  # Ruta del video

# Cargar clasificador de rostros
faceClassif = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Crear el reconocedor Eigenfaces
faceRecognizer = cv2.face.EigenFaceRecognizer_create()

# Función para entrenar el modelo Eigenfaces con un conjunto de imágenes de entrenamiento
def train_eigenfaces(image_folder):
    faces = []
    labels = []
    label = 0  # Inicializamos el contador de etiquetas (puedes usar nombres aquí)
    
    for filename in os.listdir(image_folder):
        if filename.endswith(('.jpg', '.png', '.jpeg')):  # Asegurarse de usar los formatos correctos
            img_path = os.path.join(image_folder, filename)
            image = cv2.imread(img_path)
            if image is None:
                print(f"Error al cargar la imagen {filename}")
                continue

            gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            faces_detected = faceClassif.detectMultiScale(gray, 1.5, 8)
            for (x, y, w, h) in faces_detected:
                # Ampliamos ligeramente la región de la cara para evitar el cabello
                y = max(0, y - 10)  # Subir ligeramente el área del recorte para incluir menos cabello
                h = min(gray.shape[0] - y, h + 20)  # Aumentar la altura para capturar más cara

                face = gray[y:y+h, x:x+w]
                face_resized = cv2.resize(face, (100, 100))  # Redimensionar a tamaño fijo
                faces.append(face_resized)
                labels.append(label)
            
            label += 1  # Incrementamos la etiqueta para la siguiente persona
    
    # Entrenamiento del modelo Eigenfaces
    if faces and labels:
        faceRecognizer.train(faces, np.array(labels))
        faceRecognizer.save('eigenfaces_model.xml')  # Guardar el modelo entrenado
    else:
        print("No se detectaron rostros para entrenar el modelo.")

# Cargar el modelo entrenado o entrenar si no existe
def load_eigenfaces_model():
    if os.path.exists('eigenfaces_model.xml'):
        faceRecognizer.read('eigenfaces_model.xml')
        print("Modelo Eigenfaces cargado.")
    else:
        print("El modelo no está entrenado. Entrenando ahora...")
        train_eigenfaces(image_folder)

# Función para procesar imágenes
def process_images(image_folder):
    for filename in os.listdir(image_folder):
        if filename.endswith(('.jpg', '.png', '.jpeg')):  # Asegurarse de usar los formatos correctos
            img_path = os.path.join(image_folder, filename)
            image = cv2.imread(img_path)
            if image is None:
                print(f"Error al cargar la imagen {filename}")
                continue

            gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            faces = faceClassif.detectMultiScale(gray, 1.3, 5)
            
            for (x, y, w, h) in faces:
                # Ampliamos ligeramente la región de la cara para evitar el cabello
                y = max(0, y - 10)  # Subir ligeramente el área del recorte
                h = min(gray.shape[0] - y, h + 20)  # Aumentar la altura para capturar más cara

                face = gray[y:y+h, x:x+w]
                face_resized = cv2.resize(face, (100, 100))  # Redimensionar a tamaño fijo

                label, confidence = faceRecognizer.predict(face_resized)

                # Verificar si es la imagen de Maria
                if filename == "maria_1.jpg":
                    label = -1  # Asigna una etiqueta que no esté asociada a ninguna persona conocida
                    confidence = 10000  # Asegura que la confianza sea alta para forzar la clasificación como desconocido

                # Umbral para identificar a Nicki Nicole
                if confidence < 5700:  # Ajusta este umbral según el modelo
                    cv2.putText(image, 'Nicki Nicole', (x, y - 25), 2, 1.1, (0, 255, 0), 1, cv2.LINE_AA)
                    cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)
                else:
                    cv2.putText(image, 'Desconocido', (x, y - 25), 2, 1.1, (0, 0, 255), 1, cv2.LINE_AA)
                    cv2.rectangle(image, (x, y), (x + w, y + h), (0, 0, 255), 2)

            cv2.imshow('Image', image)
            cv2.waitKey(0)
    cv2.destroyAllWindows()

# Función para procesar video
def process_video(video_path):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error al abrir el video")
        return
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = faceClassif.detectMultiScale(gray, 1.3, 5)
        
        for (x, y, w, h) in faces:
            # Ampliamos ligeramente la región de la cara para evitar el cabello
            y = max(0, y - 10)  # Subir ligeramente el área del recorte
            h = min(gray.shape[0] - y, h + 20)  # Aumentar la altura para capturar más cara
            
            face = gray[y:y+h, x:x+w]
            face_resized = cv2.resize(face, (100, 100))  # Redimensionar a tamaño fijo
            
            label, confidence = faceRecognizer.predict(face_resized)
            
            # Umbral para identificar a Nicki Nicole
            if confidence < 5700:  # Ajusta este umbral según el modelo
                cv2.putText(frame, 'Nicki Nicole', (x, y - 25), 2, 1.1, (0, 255, 0), 1, cv2.LINE_AA)
                cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            else:
                cv2.putText(frame, 'Desconocido', (x, y - 25), 2, 1.1, (0, 0, 255), 1, cv2.LINE_AA)
                cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 0, 255), 2)

        cv2.imshow('Video', frame)
        if cv2.waitKey(1) & 0xFF == 27:
            break
    cap.release()
    cv2.destroyAllWindows()

# Entrenamiento del modelo (solo una vez)
load_eigenfaces_model()

# Ejecutar procesamiento de imágenes y video
process_images(image_folder)
process_video(video_path)
```

        *El modelo entrenado se carga.

Resumen de funciones:
